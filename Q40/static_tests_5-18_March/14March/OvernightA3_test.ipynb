```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
```

```python
df = pd.read_excel("/Users/eagmurray/Documents/Data Analysis/Q40/A3_TEST.xlsx")
```

```python
df["field"].unique()
```

```python
fields = df["field"].unique()

arrays = {
    f"{field}_array": df[df["field"] == field][["time","value"]].to_numpy()
    for field in fields
}
```

```python
time_lat = pd.to_datetime(arrays["latitude_array"][:,0])
lat = pd.to_numeric(arrays["latitude_array"][:,1])

time_lon = pd.to_datetime(arrays["longitude_array"][:,0])
lon = pd.to_numeric(arrays["longitude_array"][:,1])
```

```python
idx = np.argsort(time_lat)
time_lat_sorted = time_lat[idx]
lat_sorted = lat[idx]

idx = np.argsort(time_lon)
time_lon_sorted = time_lon[idx]
lon_sorted = lat[idx]
```

```python
time_lat_sorted
```

```python
fig, (ax1,ax2) = plt.subplots(2,1,figsize=(10,6))
ax1.plot(time_lat_sorted,lat_sorted,scaley=True)
ax1.set_xlabel("Time (UTC datetime)")
ax1.set_ylabel("Latitude")

ax2.plot(time_lon_sorted,lon_sorted,scaley=True)
ax2.set_xlabel("Time (UTC datetime)")
ax2.set_ylabel("Longitude")
```
