```python
#importing required python libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
import asn1tools
```

```python
#loading the .jsonl log files
df_m10_decoded = pd.read_json("PATH/UBX_decoded_sanitised.jsol",lines=True)
df_q40_decoded = pd.read_json("PATH/q40_decoded_sanitised.jsonl",lines=True)
```

```python
#converting times to date-time for plotting
df_m10_decoded["ts"] = pd.to_datetime(df_m10_decoded["ts"])
df_q40_decoded["ts"] = pd.to_datetime(df_q40_decoded["ts"])
```

```python
print(df_q40_decoded.columns)
```

```python
#Changing height units mm -> m
df_m10_decoded["height"]= df_m10_decoded["height"]/100
df_m10_decoded["hMSL"]= df_m10_decoded["hMSL"]/100
```

```python
#plotting the data
fig,(ax1,ax2,ax3) = plt.subplots(3,1,figsize=(12,6),dpi=300)
ax1.plot(df_m10_decoded["ts"],df_m10_decoded["lon"],label="COTS")
ax1.plot(df_q40_decoded["ts"],df_q40_decoded["lon"],label="Q40")
ax1.legend()
ax1.set_title("Longitude - Overnight 17 March")
ax1.set_ylabel("Longitude")
ax1.set_xlabel("UTC Datetime")

ax2.plot(df_m10_decoded["ts"],df_m10_decoded["lat"],label="COTS")
ax2.plot(df_q40_decoded["ts"],df_q40_decoded["lat"],label="Q40")
ax2.legend()
ax2.set_title("Latitude - Overnight 17 March")
ax2.set_ylabel("Latitude")
ax2.set_xlabel("UTC Datetime")

ax3.plot(df_m10_decoded["ts"],df_m10_decoded["hMSL"],label="COTS")
ax3.plot(df_q40_decoded["ts"],df_q40_decoded["alt"],label="Q40")
ax3.legend()
ax3.set_title("Altitude - Overnight 17 March")
ax3.set_ylabel("Altitude (m)")
ax3.set_xlabel("UTC Datetime")

plt.tight_layout(pad=0.4, w_pad=0.5, h_pad=1.0)
```
