```python
#importing required python libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
import asn1tools
```

```python
#loading the .jsonl log files
df_q40_decoded = pd.read_json("/Users/eagmurray/Projects/DataAnalysis/Q40/18March/Q40_1718March/decoded_log.jsonl",lines=True)
df_m10_decoded = pd.read_json("/Users/eagmurray/Projects/DataAnalysis/Q40/18March/M10_1718March/decoded_log.jsonl",lines=True)
```

```python
#converting times to date-time for plotting
df_m10_decoded["ts"] = pd.to_datetime(df_m10_decoded["ts"])
df_q40_decoded["ts"] = pd.to_datetime(df_q40_decoded["ts"])
```

```python
df_q40_decoded["lat"][3]
```

```python
delta_lat = df_q40_decoded["lat"]-df_q40_decoded["lat"][0]
delta_lon = df_q40_decoded["lon"]-df_q40_decoded["lon"][0]
```

```python
df_q40_decoded["DoP"][0]
```

```python
hDOP=np.zeros(len(df_q40_decoded))
vDOP=np.zeros(len(df_q40_decoded))
gDOP=np.zeros(len(df_q40_decoded))
tDOP=np.zeros(len(df_q40_decoded))
pDOP=np.zeros(len(df_q40_decoded))

for i in range(len(df_q40_decoded)):
    hDOP[i] = df_q40_decoded["DoP"][i]["horizontalDilutionOfPrecision"]
    vDOP[i] = df_q40_decoded["DoP"][i]["verticalDilutionOfPrecision"]
    gDOP[i] = df_q40_decoded["DoP"][i]["geometricDilutionOfPrecision"]
    tDOP[i] = df_q40_decoded["DoP"][i]["timeDilutionOfPrecision"]
    pDOP[i] = df_q40_decoded["DoP"][i]["positionDilutionOfPrecision"]
```

```python
lat_pdop_cov = np.cov(delta_lat,pDOP)
lat_hdop_cov = np.cov(delta_lat,hDOP)
lat_vdop_cov = np.cov(delta_lat,vDOP)
lat_gdop_cov = np.cov(delta_lat,gDOP)
lat_tdop_cov = np.cov(delta_lat,tDOP)

lon_pdop_cov = np.cov(delta_lon,pDOP)
lon_hdop_cov = np.cov(delta_lon,hDOP)
lon_vdop_cov = np.cov(delta_lon,vDOP)
lon_gdop_cov = np.cov(delta_lon,gDOP)
lon_tdop_cov = np.cov(delta_lon,tDOP)
```

```python
print(f"{lat_pdop_cov}\n")
print(f"{lat_hdop_cov}\n")
print(f"{lat_vdop_cov}\n")
print(f"{lat_gdop_cov}\n")
print(f"{lat_tdop_cov}\n")

print(f"{lon_pdop_cov}\n")
print(f"{lon_hdop_cov}\n")
print(f"{lon_vdop_cov}\n")
print(f"{lon_gdop_cov}\n")
print(f"{lon_tdop_cov}\n")
```
