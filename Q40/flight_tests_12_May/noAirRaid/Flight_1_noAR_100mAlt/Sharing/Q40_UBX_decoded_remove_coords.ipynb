```python
# Confirmed offsets
lat_offset = 16
lon_offset = 24

def sanitise_raw_hex(hex_string: str) -> str:
    payload = bytearray(bytes.fromhex(hex_string))

    payload[lat_offset:lat_offset+8] = struct.pack("<d", 0.0)
    payload[lon_offset:lon_offset+8] = struct.pack("<d", 0.0)

    return payload.hex()
```

```python
# Make a copy so your original df stays untouched
df_q40_shared = df_q40_raw.copy()

# Apply sanitisation to every row
df_q40_shared["raw"] = df_q40_shared["raw"].apply(sanitise_raw_hex)

df_q40_shared.to_json("q40_raw_log_sanitised.jsonl", orient="records", lines=True)
```
