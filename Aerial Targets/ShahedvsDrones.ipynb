```python
m,b = np.polyfit(x,y,1)
print(f"Slope is {m} and intercept is {b}")
```

```python
plt.scatter(x,y,label = 'Data points')
plt.plot(x, m*x + b, color = "red", label = 'Linear fit')
plt.xlabel("Dates")
plt.ylabel("Ratio (%)")
plt.title("Shahed ratio ")
```

```python
int = 0.579710145
def func_no_intercept(x,m):
    return m*x + int

popt,pcov = sp.optimize.curve_fit(func_no_intercept,x,y,p0=0.0003268902518678984)
slope = popt[0]
```

```python
print(f"Fitted slope (m): {slope}")
print(f"Forced intercept (b): {int}")
```
