```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_189 = pd.read_csv("~/Downloads/2026-04-30_10-16-49.csv",header=None)
df_189 = df_189.iloc[:, 1:-58]
df_144 = pd.read_csv("~/Downloads/2026-04-30_10-16-21.csv", header=None)
df_144 = df_144.iloc[:,1:-58]
```

```python
base = [
    "sn","time_s","time_ms","lat","lon","temp","last_cal_temp","cal_count",
    "input_current_mA","input_voltage_mV","freq_MHz","freq_kHz","product_variant",
    "antenna_gain_dB","rx_frames_ok","rx_frame_errors","rx_crc_errors",
    "frames_relayed","tx_frames_ok","tx_drop_overrun","tx_drop_no_link",
    "tx_drop_mac_busy","tx_missed_ack","tx_bandwidth_kbps","rx_bandwidth_kbps",
    "num_sites"
]

site = [
    "site_id","tx_modulation","tx_bw","rx_bw","selected_radio_sn","seconds_since_rx",
    "missing_rx_seq","last_rx_seq","last_tx_seq","last_immediate_rx",
    "last_immediate_tx","mean_rx_margin","max_rx_margin","min_rx_margin",
    "modulation_mode","beam_x","beam_y","time_since_distance","distance_m",
    "time_since_ranging","ranging_level","time_quality","time_offset",
    "antenna_gain_site","tx_power","nlos_loss","fec_corrected","corr_peak","corr_floor"
]
```

```python
headers = base + [
    f"site{i}_{name}"
    for i in range(1, (df_189.shape[1] - len(base)) // len(site) + 1)
    for name in site
]
headers += [f"extra_{i}" for i in range(len(headers) + 1, df_189.shape[1] + 1)]
df_189.columns = headers
df_144.columns = headers
```
